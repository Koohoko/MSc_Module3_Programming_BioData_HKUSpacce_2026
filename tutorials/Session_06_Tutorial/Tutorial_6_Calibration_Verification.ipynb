## Step 1: Load the Data

**Run this cell** to create the linearity dataset.

```r
# Linearity data: 6 concentrations × 3 replicates
linearity_data <- data.frame(
  Concentration = rep(c(0, 25, 50, 100, 200, 400), each = 3),
  Signal = c(
    0.052, 0.048, 0.055,
    0.245, 0.251, 0.238,
    0.478, 0.485, 0.472,
    0.952, 0.965, 0.948,
    1.875, 1.892, 1.868,
    3.425, 3.458, 3.412
  )
)

head(linearity_data, 9)
```

---

## Step 2: Fit the Linear Model

**Task:** Fill in the blank to fit a linear model (Signal vs Concentration).

```r
# TODO: Fit linear model
# Formula: Signal ~ Concentration
model <- lm(... ~ ..., data = linearity_data)  # <-- FILL IN THE BLANKS

# View summary
summary(model)
```

---

## Step 3: Extract Key Values

**Task:** Fill in the blanks to extract slope, intercept, and R².

```r
# TODO: Extract values from the model
intercept <- coef(model)[...]  # <-- FILL IN (1 or 2?)
slope <- coef(model)[...]      # <-- FILL IN (1 or 2?)
r_squared <- summary(model)$r.squared

# Print the calibration equation
cat("Equation: Signal =", round(slope, 5), "× Concentration +", round(intercept, 4), "\n")
cat("R² =", round(r_squared, 4))
```

**Expected:** R² ≈ 0.9987

---

## Step 4: Create the Calibration Plot

**Run this cell** to see the calibration curve.

```r
# Calibration plot with ggplot2
ggplot(linearity_data, aes(x = Concentration, y = Signal)) +
  geom_point(size = 3, color = "steelblue") +
  geom_smooth(method = "lm", se = TRUE, color = "red") +
  labs(
    title = "Calibration Curve",
    subtitle = paste0("R² = ", round(r_squared, 4)),
    x = "Concentration (mg/dL)",
    y = "Signal (OD)"
  ) +
  theme_minimal()
```

## Step 6: Pass/Fail Decision

**Run this cell** to see the verification result.

```r
# Acceptance check: R² ≥ 0.99
threshold <- 0.99

cat("R² =", round(r_squared, 4), "\n")
cat("Criterion: R² ≥", threshold, "\n\n")

if (r_squared >= threshold) {
  cat("✅ PASS: Linearity is acceptable.\n")
} else {
  cat("⚠️ FAIL: Linearity does not meet criteria.\n")
}
```
