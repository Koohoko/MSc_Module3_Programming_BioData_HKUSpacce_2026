## Step 3: Explore the Sequences

```r
# How many sequences?
cat("Number of sequences:", length(sequences), "\n")

# What are their names?
cat("Variant names:\n")
print(names(sequences))

# How long are they?
cat("\nSequence lengths (bp):\n")
print(width(sequences))
```

## Step 4: Calculate GC Content

**Task:** Fill in the blank to calculate GC content for all sequences.

```r
# TODO: Calculate GC content
# Hint: letterFrequency(sequences, letters = c("G", "C"), as.prob = TRUE)

gc_freq <- letterFrequency(sequences, letters = c("...", "..."), as.prob = TRUE)  # <-- FILL IN
gc_percent <- rowSums(gc_freq) * 100

# Create results table
results <- data.frame(
  Variant = names(sequences),
  Length_bp = width(sequences),
  GC_Content = round(gc_percent, 1)
)

print(results)
```

## Step 6: Compare a Position

**Run this cell** to see nucleotide at a specific position.

```r
# Extract nucleotide at position 50 for each variant
position <- 50

cat("Nucleotide at position", position, ":\n")
for (i in 1:length(sequences)) {
  nuc <- as.character(subseq(sequences[[i]], start = position, end = position))
  cat(names(sequences)[i], ":", nuc, "\n")
}
```
