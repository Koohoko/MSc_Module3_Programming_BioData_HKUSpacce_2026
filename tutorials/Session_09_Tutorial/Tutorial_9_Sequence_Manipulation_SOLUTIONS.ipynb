```r
# Create FASTA
fasta_content <- ">Wuhan-Hu-1
ATGTTTGTTTTTCTTGTTTTATTGCCACTAGTCTCTAGTCAGTGTGTTAATCTTACAACCAGAACTCAATTACCCCCTGCATACACTAATTCTTTCACACGTGGTGTTTATTACCCTGAC
>Alpha_B.1.1.7
ATGTTTGTTTTTCTTGTTTTATTGCCACTAGTCTCTAGTCAGTGTGTTAATCTTACAACCAGAACTCAATTACCCCCTGCATACACTAATTCTTTCACACGTGGTGTTTATTACCCTGAC
>Delta_B.1.617.2
ATGTTTGTTTTTCTTGTTTTATTGCCACTAGTCTCTAGTCAGTGTGTTAATCTTACAACCAGAACTCAATTACCCCCTGCATACACTAATTCTTTCACACGTGGTGTTTATTACCCTGAC
>Omicron_BA.1
ATGTTTGTTTTTCTTGTTTTATTGCCACTAGTCTCTAGTCAGTGTGTTAATCTTACAACCAGAACTCAATTACCCCCTGCATACACTAATTCTTTCACACGTGGTGTTTATTACCCTGAC"

writeLines(fasta_content, "sars_cov2_variants.fasta")
```

```r
# SOLUTION: Step 2
sequences <- readDNAStringSet("sars_cov2_variants.fasta")
print(sequences)
```

```r
# SOLUTION: Step 4 - GC Content
gc_freq <- letterFrequency(sequences, letters = c("G", "C"), as.prob = TRUE)
gc_percent <- rowSums(gc_freq) * 100

results <- data.frame(
  Variant = names(sequences),
  Length_bp = width(sequences),
  GC_Content = round(gc_percent, 1)
)
print(results)
```

```r
# SOLUTION: Step 5 - Translate
proteins <- translate(sequences)
print(proteins)
```
