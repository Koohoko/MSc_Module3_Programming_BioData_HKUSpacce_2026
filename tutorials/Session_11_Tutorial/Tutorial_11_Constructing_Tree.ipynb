## Step 1: Load the Sequences

**Run this cell** to create the sequence data.

```r
# Outbreak sequences
seqs <- DNAStringSet(c(
  "Wuhan_Ref" = "ATGTTTGTTTTTCTTGTTTTATTGCCACTAGTCTCTAGTCAGTGTGTTAATCTTACAACC",
  "Alpha_Ref" = "ATGTTTGTTTTTCTTGTTTTATTGCCACTAGTCTCTAGTCAGTGTGTTAATCTTACAACC",
  "Delta_Ref" = "ATGTTTGTTTTTCTTGTTTTATTGCCACTAGTCTCTAGTCAGTGTGTTAATCTTACACCC",
  "Omicron_Ref" = "ATGTTTGTTTTTCTTGTTTTATTGCCACTAATCTCTAGTCAGTGTGTTAATCTTACAACC",
  "Patient_1_HK" = "ATGTTTGTTTTTCTTGTTTTATTGCCACTAATCTCTAGTCAGTGTGTTAATCTTACAACT",
  "Patient_2_HK" = "ATGTTTGTTTTTCTTGTTTTATTGCCACTAATCTCTAGTCAGTGTGTTAATCTTACAACC",
  "Patient_3_HK" = "ATGTTTGTTTTTCTTGTTTTATTGCCACTAATCTCTAGTCAGTGTGTTAATCTTACAACC",
  "Patient_4_SZ" = "ATGTTTGTTTTTCTTGTTTTATTGCCACTAGTCTCTAGTCAGTGTGTTAATCTTACACCC",
  "Patient_5_SZ" = "ATGTTTGTTTTTCTTGTTTTATTGCCACTAGTCTCTAGTCAGTGTGTTAATCTTACACCT"
))

cat("Loaded", length(seqs), "sequences")
```

---

## Step 2: Perform Multiple Sequence Alignment

**Task:** Fill in the blank to align the sequences.

```r
# TODO: Align sequences using ClustalW
alignment <- msa(seqs, method = "...")  # <-- FILL IN (ClustalW)

# View alignment
print(alignment, show = "complete")
```

---

## Step 3: Convert to DNAbin Format

```r
# Convert for ape
aligned_seqs <- as(alignment, "DNAStringSet")
dna_bin <- as.DNAbin(aligned_seqs)

cat("Converted", length(dna_bin), "sequences to DNAbin format")
```

---

## Step 4: Calculate Distance Matrix

**Task:** Fill in the blank to calculate distances.

```r
# TODO: Calculate genetic distance
dist_matrix <- dist.dna(dna_bin, model = "...")  # <-- FILL IN (raw)

# View as matrix
round(as.matrix(dist_matrix), 4)
```

---

## Step 5: Build the Tree

**Task:** Fill in the blank to build an NJ tree.

```r
# TODO: Build Neighbor-Joining tree
tree <- ...(dist_matrix)  # <-- FILL IN (nj function)

# Plot
plot(tree, main = "Outbreak Phylogenetic Tree")
add.scale.bar()
```

---

## Step 6: Root the Tree

```r
# Root using Wuhan as outgroup
rooted_tree <- root(tree, outgroup = "Wuhan_Ref", resolve.root = TRUE)

plot(rooted_tree, main = "Rooted Tree (Wuhan outgroup)")
add.scale.bar()
```
