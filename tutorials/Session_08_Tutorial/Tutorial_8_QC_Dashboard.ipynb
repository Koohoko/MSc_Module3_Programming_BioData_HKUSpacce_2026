You are the QC officer reviewing **60 days of Creatinine QC data**. Something changed around day 31 — your task is to visualize and detect the problem.

## Step 1: Load the Data

**Run this cell** to load the QC data.

```r
# 60 days of creatinine QC data
qc_data <- data.frame(
  Day = 1:60,
  Creatinine = c(
    1.02, 0.98, 1.05, 0.96, 1.01, 1.03, 0.99, 1.04, 0.97, 1.00,
    1.02, 0.98, 1.06, 0.95, 1.01, 1.03, 0.99, 1.02, 0.97, 1.00,
    1.04, 0.98, 1.05, 0.96, 1.02, 1.00, 1.03, 0.99, 1.01, 0.97,
    1.12, 1.15, 1.08, 1.18, 1.10, 1.14, 1.09, 1.16, 1.12, 1.11,
    1.17, 1.08, 1.13, 1.19, 1.10, 1.15, 1.12, 1.14, 1.09, 1.16,
    1.22, 1.11, 1.18, 1.13, 1.15, 1.10, 1.17, 1.12, 1.14, 1.19
  )
)

cat("Loaded", nrow(qc_data), "days of QC data\n")
head(qc_data)
```

---

## Step 2: Calculate Baseline Statistics

**Task:** Calculate the mean and SD from the **first 20 days** (baseline period).

```r
# TODO: Calculate baseline mean and SD from days 1-20
baseline <- qc_data$Creatinine[1:...]  # <-- FILL IN (how many days?)

baseline_mean <- mean(baseline)
baseline_sd <- sd(baseline)

cat("Baseline Mean:", round(baseline_mean, 3), "mg/dL\n")
cat("Baseline SD:", round(baseline_sd, 3), "mg/dL")
```

**Expected:** Mean ≈ 1.00, SD ≈ 0.03

---

## Step 3: Calculate Control Limits

**Run this cell** to calculate the control limits.

```r
# Control limits
upper_3sd <- baseline_mean + 3 * baseline_sd
lower_3sd <- baseline_mean - 3 * baseline_sd
upper_2sd <- baseline_mean + 2 * baseline_sd
lower_2sd <- baseline_mean - 2 * baseline_sd

cat("=== Control Limits ===\n")
cat("+3SD:", round(upper_3sd, 3), "\n")
cat("+2SD:", round(upper_2sd, 3), "\n")
cat("Mean:", round(baseline_mean, 3), "\n")
cat("-2SD:", round(lower_2sd, 3), "\n")
cat("-3SD:", round(lower_3sd, 3))
```

---

## Step 4: Create the Levey-Jennings Chart

**Task:** Fill in the blank to add the +3SD control line.

```r
# TODO: Create LJ chart
ggplot(qc_data, aes(x = Day, y = Creatinine)) +
  # Control lines
  geom_hline(yintercept = baseline_mean, color = "darkgreen", linewidth = 1) +
  geom_hline(yintercept = upper_2sd, color = "orange", linetype = "dashed") +
  geom_hline(yintercept = lower_2sd, color = "orange", linetype = "dashed") +
  geom_hline(yintercept = ..., color = "red", linewidth = 1) +  # <-- FILL IN (upper 3SD limit)
  geom_hline(yintercept = lower_3sd, color = "red", linewidth = 1) +
  
  # Data
  geom_line(color = "steelblue") +
  geom_point(size = 2, color = "steelblue") +
  
  labs(title = "Levey-Jennings Chart: Creatinine QC",
       x = "Day", y = "Creatinine (mg/dL)") +
  theme_minimal()
```

---

## Step 5: Identify Violations

**Task:** Create a column to flag 1:3s violations.

```r
# TODO: Flag 1:3s violations
# A violation occurs when the value is more than 3 SD from the mean
qc_data$Violation <- abs(qc_data$Creatinine - baseline_mean) > ... * baseline_sd  # <-- FILL IN (3)

# Count violations
cat("Number of 1:3s violations:", sum(qc_data$Violation))
```

---

## Step 6: Visualize with Violations Highlighted

**Run this cell** to see violations in red.

```r
# LJ Chart with violations
ggplot(qc_data, aes(x = Day, y = Creatinine)) +
  geom_hline(yintercept = baseline_mean, color = "darkgreen", linewidth = 1) +
  geom_hline(yintercept = upper_3sd, color = "red", linewidth = 1) +
  geom_hline(yintercept = lower_3sd, color = "red", linewidth = 1) +
  
  # Mark shift point
  geom_vline(xintercept = 30.5, color = "purple", linetype = "dashed") +
  
  # Data line
  geom_line(color = "steelblue") +
  
  # Normal points
  geom_point(data = qc_data[!qc_data$Violation, ], size = 2, color = "steelblue") +
  
  # Violation points (RED)
  geom_point(data = qc_data[qc_data$Violation, ], size = 3, color = "red", shape = 17) +
  
  annotate("text", x = 45, y = upper_3sd + 0.02, label = "SHIFT!", color = "red", fontface = "bold") +
  
  labs(title = "QC Dashboard: Creatinine",
       subtitle = "Systematic shift detected at Day 31",
       x = "Day", y = "Creatinine (mg/dL)") +
  theme_minimal()
```
