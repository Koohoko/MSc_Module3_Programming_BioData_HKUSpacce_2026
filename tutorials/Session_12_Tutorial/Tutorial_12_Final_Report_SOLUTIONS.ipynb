```r
# SETUP - Run this cell
options(repos = c(CRAN = "https://cloud.r-project.org"))
library(ggplot2)

cat("✅ Ready!")
```

---

## Step 1: Create Project Structure

Pro Tip: Always separate **Data** (Input), **Scripts** (Code), and **Results** (Output).

**Task:** Use `dir.create()` to make three folders: `my_data`, `my_scripts`, `my_results`.

```r
# TODO: Create directories
dir.create("my_data")
dir.create("my_scripts")
dir.create("my_results")

# Verify
list.dirs(recursive = FALSE)
```

---

## Step 2: Create Dummy Data

Let's create a CSV file and save it into the `my_data` folder.

```r
# Create a simple dataset
my_data <- data.frame(
    Sample = paste0("S", 1:5),
    Value = c(10, 23, 15, 30, 22)
)

# TODO: Save to 'my_data' folder as 'experiment.csv'
write.csv(my_data, "my_data/experiment.csv", row.names = FALSE)

cat("Data saved!")
```

---

## Step 3: Write an Analysis Script

Instead of running code cells one by one, we will write a script file (`.R`).

**Task:** Create a script that:
1. Reads `my_data/experiment.csv`
2. Makes a bar plot
3. Saves it to `my_results/plot.png`

```r
# We write the script content to a text variable first
script_content <- '
print("Starting analysis...")
library(ggplot2)

# 1. Read Data
df <- read.csv("my_data/experiment.csv")

# 2. Plot
p <- ggplot(df, aes(x = Sample, y = Value)) + 
     geom_col(fill = "steelblue") +
     theme_minimal() +
     ggtitle("Automated Plot")

# 3. Save Result
ggsave("my_results/plot.png", p)
print("Analysis complete! Plot saved.")
'

# TODO: Write this content to 'my_scripts/run_analysis.R'
writeLines(script_content, "my_scripts/run_analysis.R")
```

---

## Step 4: Execute the Pipeline

Now, act as the "User" running the analysis. You just need to run **one command** to trigger the whole process.

```r
# TODO: Run the script using source()
source("my_scripts/run_analysis.R")

# Verify output exists
file.exists("my_results/plot.png")
```
