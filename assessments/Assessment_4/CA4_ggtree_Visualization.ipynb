# Section 1: Generate Your Dataset (10 marks)

```r
# ============================================
# SETUP - Run first
# ============================================
options(repos = c(CRAN = "https://cloud.r-project.org"))

if (!requireNamespace("BiocManager", quietly = TRUE)) install.packages("BiocManager")
if (!requireNamespace("ggtree", quietly = TRUE)) BiocManager::install("ggtree", update = FALSE, ask = FALSE)
if (!requireNamespace("ape", quietly = TRUE)) install.packages("ape")

library(ggtree)
library(ape)
library(ggplot2)

cat("✅ Packages loaded!")
```

```r
# ============================================
# ENTER YOUR STUDENT ID HERE
# ============================================

student_id <- "YOUR_ID_HERE"  # <-- REPLACE

# ============================================
# DO NOT MODIFY BELOW
# ============================================

id_numeric <- as.numeric(gsub("[^0-9]", "", student_id))
if (is.na(id_numeric) || id_numeric == 0) {
  stop("❌ Please enter a valid Student ID!")
}
set.seed(id_numeric)

# Generate unique tree
n_tips <- sample(8:12, 1)
tree <- rtree(n_tips)
tree$tip.label <- paste0("Sample_", 1:n_tips)

# Generate metadata
locations <- sample(c("Hong_Kong", "Shenzhen", "Guangzhou"), n_tips, replace = TRUE)
variants <- sample(c("Alpha", "Delta", "Omicron"), n_tips, replace = TRUE)

metadata <- data.frame(
  label = tree$tip.label,
  Location = locations,
  Variant = variants
)

cat("✅ Dataset generated for Student ID:", student_id, "\n")
cat("Tree with", n_tips, "tips\n\n")
print(metadata)
```

# Section 5: Final Figure with Scale Bar (15 marks)

**Task:** Create final publication figure and save as PNG.

```r
# TODO: Create final figure with scale bar
final_plot <- ggtree(tree, layout = "circular") %<+% metadata +
  geom_tiplab(aes(color = Location, angle = angle), hjust = -0.1, size = 4) +
  geom_tippoint(aes(color = Location), size = 3) +
  ...() +  # <-- FILL IN (add scale bar function)
  scale_color_brewer(palette = "Set1") +
  labs(title = paste("Phylogeny - Student:", student_id))

print(final_plot)
```

```r
# Save as PNG
filename <- paste0("tree_", student_id, ".png")
ggsave(filename, final_plot, width = 10, height = 8, dpi = 300)
cat("Saved to:", filename)
```
