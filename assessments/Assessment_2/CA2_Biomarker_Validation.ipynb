# Section 1: Generate Your Dataset (10 marks)

```r
# ============================================
# ENTER YOUR STUDENT ID HERE
# ============================================

student_id <- "YOUR_ID_HERE"  # <-- REPLACE

# ============================================
# DO NOT MODIFY BELOW
# ============================================

options(repos = c(CRAN = "https://cloud.r-project.org"))
if (!requireNamespace("pROC", quietly = TRUE)) install.packages("pROC")
library(pROC)

id_numeric <- as.numeric(gsub("[^0-9]", "", student_id))
if (is.na(id_numeric) || id_numeric == 0) {
  stop("❌ Please enter a valid Student ID!")
}
set.seed(id_numeric)

# Generate biomarker data
n_cases <- 30
n_controls <- 30

biomarker <- data.frame(
  SampleID = paste0("S", sprintf("%03d", 1:(n_cases + n_controls))),
  Level = c(
    rnorm(n_cases, mean = runif(1, 6.5, 7.5), sd = runif(1, 1.5, 2.5)),
    rnorm(n_controls, mean = runif(1, 3.5, 4.5), sd = runif(1, 1.2, 2.0))
  ),
  Group = c(rep("Case", n_cases), rep("Control", n_controls))
)

# Add binary outcome for ROC
biomarker$Outcome <- ifelse(biomarker$Group == "Case", 1, 0)

cat("✅ Dataset generated for Student ID:", student_id, "\n\n")
cat("Samples:", nrow(biomarker), "(30 Cases, 30 Controls)\n\n")
head(biomarker, 10)
```
